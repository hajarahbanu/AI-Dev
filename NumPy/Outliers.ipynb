```python
import numpy as np
```

```python
arr = np.array([2,3,4,6,7,8,9,12,13,16,17,23,25,34,37,201])
arr
```

```python
q1 = np.percentile(arr,25)
q1
```

```python
q3 = np.percentile(arr,75)
q3
```

```python
IQR = q3 - q1
IQR
```

```python
uf = q3 + (1.5 * IQR)
```

```python
Lf = q1 - (1.5 * IQR)
```

```python
l = []
for i in arr:
    if i >= Lf and i <= uf:
        l.append(i)
arr2 = np.array(l)
arr2
```

```python
import seaborn as sns
```

```python
# boxplot represents the 5 number summary
# it also shows the outlier in the data

sns.boxplot(x = arr)
```

```python
sns.boxplot(x = arr2)
```
