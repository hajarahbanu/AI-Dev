```python
import numpy as np
```

```python
arr = np.arange(1,21)
arr
```

```python
# indexing
arr[6]
```

```python
# slicing
arr[:5]
```

```python
arr[:]
```

```python
arr[2::2]
# it starts with index no. 2 and keeps jumping 2 elements
```

```python
# two-dimensional array indexing
arr2d = np.arange(1,31).reshape(6,5)
arr2d
```

```python
arr2d[0]
```

```python
arr2d[0,1]
```

```python
arr2d[0:2,1:3]
# from 0th row to 1st row, and 1st coloumn to 2nd coloumns
```

```python
# arr2d[3:6, 3:5]
arr2d[3:,3:]
```

```python
# boolean indexing
arrbool = np.arange(1,21)
arrbool
```

```python
bool_index = arrbool % 2 == 0
bool_index
```

```python
arrbool = arrbool[bool_index]
# using bool index to store only even elements in an array
arrbool
```
