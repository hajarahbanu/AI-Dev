```python
# shallow ccopy
a = np.array([3,5,3,4,7,8])
slice = a[:3]
slice
```

```python
slice * 10 
```

```python
a
```

```python
# operations done on slice is not affected in original array , this is known as shallow copy
```

```python
b = a
```

```python
b + 10 
```

```python
a
```

```python
b[0]= 99
```

```python
b
```

```python
a
# here chaning in b has changed a as well, this is known as deep copy
```

```python
# matrix multiplication
mata = np.arange(1,5).reshape(2,2)
```

```python
mata
```

```python
matb = np.arange(5,9).reshape(2,2)
```

```python
matb
```

```python
mata * matb
```

```python
# the above multiplies based on index and not doing actual matrix multiplication
# for this numpy has "@" operator
```

```python
mata @ matb
```

```python
np.dot(mata, matb)
```

```python
mata.T
# to perform transpose
```
