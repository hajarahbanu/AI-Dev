```python
import pandas as pd
```

```python
import numpy as np
```

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
# Group by Category and calculate the sum of Sales
# how to split category
cat = df.groupby('Category')
cat
```

```python
# iterate via objects to get the actual value
for i,v in cat:
    print(i)
    print(v)
```

```python
sales_cat = df.groupby('Category')['Sales'].sum()
sales_cat
```

```python
# Group by Store and calculate the sum of Sales
sales_stores = df.groupby('Store')['Sales'].sum()
sales_stores
```

```python
# calculate sales by grouping category and store
df.groupby(['Category', 'Store'])['Sales'].sum()
```

```python
# calculate sales by grouping category and store
df.groupby(['Store', 'Category'])['Sales'].sum()
```

```python
# combined statistics on data can be performed bbytearray
df['Sales'].agg(['mean', 'median', 'min','max','sum', 'count', 'std'])
```
